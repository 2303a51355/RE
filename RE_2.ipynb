```python
import random
machines = ['A', 'B', 'C']
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0, 0, 0]

epsilon = 0.2

for i in range(10):

    r = random.random()

    if r < epsilon:
        # Exploration
        x = random.randint(0, 2)
    else:
        # Exploitation
        x = Q.index(max(Q))

    print("Play", i + 1)
    print("Selected Machine:", machines[x])

    reward = float(input("Enter reward: "))

    N[x] = N[x] + 1
    R[x] = R[x] + reward

    Q[x] = Q[x] + (reward - Q[x]) / N[x]

    print("N =", N)
    print("R =", R)
    print("Q =", Q)
    print()

print("Final Result")
print("Machine   N   R   Q")

for i in range(3):
    print(machines[i], "       ", N[i], " ", R[i], " ", round(Q[i], 2))
```
