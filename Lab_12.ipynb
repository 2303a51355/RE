```python
gamma = 0.9
theta = 0.0001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": "S0",
        "Down": "S1",
        "Left": "S0",
        "Right": "S0"
    },
    "S1": {
        "Up": "S0",
        "Down": "S1",
        "Left": "S1",
        "Right": "G"
    }
}
V = {state: 0.0 for state in states}


def reward(next_state):
    if next_state == "G":
        return 100
    return -1
print("STATE TRANSITION INFORMATION")
print("State\tAction\tNext State\tReward")

for state in transitions:
    for action in actions:
        next_state = transitions[state][action]
        r = reward(next_state)
        print(f"{state}\t{action}\t{next_state}\t\t{r}")

iteration = 0

print("\nITERATION-WISE STATE VALUES")
print("Iteration\tV(S0)\t\tV(S1)\t\tV(G)")

while True:
    iteration += 1
    new_V = V.copy()

    for state in ["S0", "S1"]:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]
            r = reward(next_state)

            q = r + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(
        abs(new_V[state] - V[state])
        for state in ["S0", "S1"]
    )

    V = new_V

    print(f"{iteration}\t\t{V['S0']:.4f}\t\t"
          f"{V['S1']:.4f}\t\t{V['G']:.4f}")

    if delta < theta:
        break
policy = {}

print("\nFINAL ACTION-VALUE TABLE")
print("State\tAction\tNext State\tQ(s,a)")

for state in ["S0", "S1"]:
    q_values = {}

    for action in actions:
        next_state = transitions[state][action]
        r = reward(next_state)

        q = r + gamma * V[next_state]
        q_values[action] = q

        print(f"{state}\t{action}\t{next_state}\t\t{q:.2f}")

    policy[state] = max(q_values, key=q_values.get)
print("\nOPTIMAL STATE VALUES")
for state in states:
    print(f"V*({state}) = {V[state]:.2f}")
print("\nOPTIMAL POLICY")
for state in policy:
    print(f"{state} -> {policy[state]}")
print("\nOPTIMAL PATH")
state = "S0"
path = [state]

while state != "G":
    action = policy[state]
    state = transitions[state][action]
    path.append(state)

print(" -> ".join(path))
```
