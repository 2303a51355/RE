```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Part B – Knowledge Initialization
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# To get the same result every time
random.seed(10)

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("-------------------------------------------")

# Part C – Modified Epsilon-Greedy Decision
for play in range(1, N_play + 1):

    r = random.random()

    # Find highest Q-value machine
    max_q = max(Q.values())
    best_machines = [m for m in machines if Q[m] == max_q]

    # Exploration
    if r < epsilon:

        # Select randomly except the highest estimated reward machine
        other_machines = [m for m in machines if m not in best_machines]

        # If all machines have equal Q initially,
        # randomly select any machine
        if len(other_machines) == 0:
            selected_machine = random.choice(machines)
        else:
            selected_machine = random.choice(other_machines)

        decision = "Exploration"

    # Exploitation
    else:
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    # Part D – Reward Collection
    print("\nPlay", play)
    print("Random number:", round(r, 3))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)

    reward = float(input("Enter reward: "))

    # Part E – Q-Value Update
    N[selected_machine] += 1
    R[selected_machine].append(reward)

    old_q = Q[selected_machine]

    Q[selected_machine] = (
        old_q + (1 / N[selected_machine]) *
        (reward - old_q)
    )

    print("Updated Q-value:",
          round(Q[selected_machine], 2))


# Part F – Display Results
print("\n\n========== FINAL RESULTS ==========")

print("\nParticipation Count (N):")
for machine in machines:
    print(machine, ":", N[machine])

print("\nRewards (R):")
for machine in machines:
    print(machine, ":", R[machine])

print("\nEstimated Q-values:")
for machine in machines:
    print(machine, ":", round(Q[machine], 2))

# Total reward
total_reward = sum(
    sum(R[machine]) for machine in machines
)

print("\nTotal Reward Received:", total_reward)

# Best machine
best_machine = max(Q, key=Q.get)

print("Best Machine:", best_machine)
print("Highest Q-value:", round(Q[best_machine], 2))
```
